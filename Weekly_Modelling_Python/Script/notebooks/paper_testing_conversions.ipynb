```python
def calculate_market_profits(df, market, stake_amount):
    df = df[df['Bet'] == 'LAY']
    US_profit = round(sum(df['Profit'] / df['Stake']), 2)
    FS_profit = round(sum((df['Profit'] / df['Stake']) * stake_amount), 2)
    print(market, "uniform stake profit:", US_profit)
    print(market, "fixed stake profit:", FS_profit)
    return US_profit, FS_profit

euro_winner_profits = calculate_market_profits(euro_winner, "Winner", euro_winner_FS)
euro_top5_profits = calculate_market_profits(euro_top5, "Top5", euro_top5_FS)
euro_top10_profits =  calculate_market_profits(euro_top10, "Top10", euro_top10_FS)
euro_top20_profits = calculate_market_profits(euro_top20, "Top20", euro_top20_FS)

pga_winner_profits = calculate_market_profits(pga_winner, "Winner", pga_winner_FS)
pga_top5_profits = calculate_market_profits(pga_top5, "Top5", pga_top5_FS)
pga_top10_profits =  calculate_market_profits(pga_top10, "Top10", pga_top10_FS)
pga_top20_profits = calculate_market_profits(pga_top20, "Top20", pga_top20_FS)
```
